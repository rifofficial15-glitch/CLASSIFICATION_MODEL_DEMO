NOW AS WE LOOK INTO THE DATA,THE DATATYPES FOR SOME SPECIFIC COLUMNS ARE INCORRECT.. WE NEED TO IDENTIFY & CORRECT IT..

```python
DT_columns = ['PACKED_CELL_VOLUME', 'WBC_COUNT', 'RBC_COUNT']
for i in DT_columns:
    print(f"{i} : {df[i].dtype}")
```

```python
# WE ARE USING FUNCTION TO CONVERT DATATYPE TO NUMERIC
def convert_text_to_numeric(df, column):
    df[column] = pd.to_numeric(df[column], errors='coerce')
#errors='coerce' tells that if we encounter a value that cannot be converted to the target datatype, force it into missing value(NAN or NaT) instead of throwing an error or leaving it unchanged..
for column in DT_columns:
    convert_text_to_numeric(df, column)
    print(f"{column} : {df[column].dtype}")
```

```python
df.info()
```

NOW WE GOING TO HANDLE MISSING VALUES..

```python
missing = df.isna().sum()
missing[missing>0].sort_values(ascending=False)
```

```python
# HERE WE ARE FILLING THE MMISSING VALUES WITH MEAN(FOR NUMERIC) & MODE(FOR CATEGORICAL)
def mean_value_imputation(df, column):
    mean_value = df[column].mean()
    df[column] = df[column].fillna(value=mean_value)

def mode_value_imputation(df, column):
    mode_value = df[column].mode()[0]
    df[column] = df[column].fillna(mode_value)
```

```python
# FILLING MISSING VALUES FOR NUMERIC COLUMNS

num_cols = [col for col in df.columns if df[col].dtype!='str']
for col_name in num_cols:
    mean_value_imputation(df, col_name)
```

```python
# FILLING MISSING VALUES FOR CATEGORICAL COLUMNS

cat_cols = [col for col in df.columns if df[col].dtype=='str']
for col_name in cat_cols:
    mode_value_imputation(df, col_name)
```

NOW WE HAVE REPLACE THE INCORRECT VALUE..

```python
df['DIABETIS_MELLITUS'] = df['DIABETIS_MELLITUS'].replace(to_replace = {' yes':'yes', '\tno':'no', '\tyes':'yes'})
df['CORONARY_ARTERY_DISEASE'] = df['CORONARY_ARTERY_DISEASE'].replace(to_replace = '\tno', value='no')
df['CLASS'] = df['CLASS'].replace(to_replace = {'ckd\t':'ckd', 'notckd':'not ckd'})

```

```python
print(f"DIABETIS_MELLITUS : - {df['DIABETIS_MELLITUS'].unique()}")
print(f"CORONARY_ARTERY_DISEASE : - {df['CORONARY_ARTERY_DISEASE'].unique()}")
print(f"CLASS : - {df['CLASS'].unique()}")
```

NOW WE ARE GOING TO DO FEATURE ENCODING..
    (I-E) MACHINE LEARNING MODEL UNDERSTANDS ONLY THE NUMBERS.. SO WE ARE GOING TO CHANGE THE CATEGORIES TO NUMERIC VALUES..

```python
df.columns
```

```python
df['CLASS'] = df['CLASS'].map({'ckd': 1, 'not ckd': 0})
df['RED_BLOOD_CELLS'] = df['RED_BLOOD_CELLS'].map({'normal': 1, 'abnormal': 0})
df['PUS_CELLS'] = df['PUS_CELLS'].map({'normal': 1, 'abnormal': 0})
df['PUS_CELL_CLUMPS'] = df['PUS_CELL_CLUMPS'].map({'present': 1, 'notpresent': 0})
df['BACTERIA'] = df['BACTERIA'].map({'present': 1, 'notpresent': 0})
df['HYPERTENSION'] = df['HYPERTENSION'].map({'yes': 1, 'no': 0})
df['DIABETIS_MELLITUS'] = df['DIABETIS_MELLITUS'].map({'yes': 1, 'no': 0})
df['CORONARY_ARTERY_DISEASE'] = df['CORONARY_ARTERY_DISEASE'].map({'yes': 1, 'no': 0})
df['APPETITE'] = df['APPETITE'].map({'good': 1, 'poor': 0})
df['PEDA_EDEMA'] = df['PEDA_EDEMA'].map({'yes': 1, 'no': 0})
df['ANAEMIA'] = df['ANAEMIA'].map({'yes': 1, 'no': 0})


```

```python
# SPLITTING TRAIN & TEST DATA
from sklearn.model_selection import train_test_split

x = df.drop("CLASS", axis=1)
y = df["CLASS"]

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.25, random_state=25)

print(f"'x_train:' shape : {x_train.shape}")
print(f"'x_test:' shape : {x_test.shape}")
```

```python
from sklearn.tree import DecisionTreeClassifier

dct = DecisionTreeClassifier()
dct.fit(x_train, y_train)
```

```python
y_pred_dct = dct.predict(x_test)
y_pred_dct
```

```python
# CLASSIFICATION METRICS SHOULD BE USED.. BEFORE THAT WE ARE GOING TO PREDICT WITH OTHER ALGORITHMS OF CLASSIFICATION
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
```

```python
# CREATE MODEL ARRAY AND APPEND ALL ALGORITHMS..
models = []
models.append(('NAIVE BAYES', GaussianNB()))
models.append(('KNN', KNeighborsClassifier(n_neighbors=8)))
models.append(('RANDOM FOREST CLASSIFIER', RandomForestClassifier()))
models.append(('DECISION TREE CLASSIFIER', DecisionTreeClassifier()))
models.append(('SVM', SVC(kernel = 'linear')))
```

```python
# NOW CREATING METRICS
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
```

```python
for name, model in models:
    print(name, model)
    print()
    model.fit(x_train, y_train)
    y_pred = model.predict(x_test)
    print("CONFUSION MATRIX:\n", confusion_matrix(y_test, y_pred))
    print('\n')
    print("ACCURACY:", accuracy_score(y_test, y_pred))
    print('\n')
    print("PRECISION:", precision_score(y_test, y_pred))
    print('\n')
    print("RECALL:", recall_score(y_test, y_pred))
    print('\n')
    print("F1 SCORE:", f1_score(y_test, y_pred))
    print('\n')
```
